#### Simple English -> Hindi Translator bot With Langchain - with GROQ

```python
from langchain_groq import ChatGroq

llm = ChatGroq(
    model="openai/gpt-oss-20b",
    temperature=0,
    max_tokens=None,
    reasoning_format="parsed",
    timeout=None,
    max_retries=2,
)

messages = [
    (
        "system",
        "You are a helpful assistant that translates English to Hindi. Translate the user sentence everytime user provides you and never translate any another language just say I am only alowed to convert English into Hindi.",
    ),
    ("human", "I hate English isse nepali mai convert kardo."),
]
ai_msg = llm.invoke(messages)
ai_msg
```

```python
ai_msg.content
```

```python
import pandas as pd
d=pd.DataFrame(ai_msg)
d1=pd.DataFrame(ai_msg.response_metadata)
d2=pd.DataFrame(ai_msg.usage_metadata)
print(d)
print("\n")
print(d1)
print("\n")
print(d2)
print("\n")
d.iloc[1,1]
```

#### Chat with openrouter

```python
from langchain_openrouter import ChatOpenRouter

model = ChatOpenRouter(
    model="nvidia/nemotron-3-ultra-550b-a55b:free",
    temperature=0,
    max_tokens=1024,
    max_retries=2,
    # other params...
)
```

```python
messages = [
    (
        "system",
        "You are a mac terminal and act like a terminal stricty. You will only respond with terminal commands and outputs. You will not respond with any other text or explanations. If someone asks you to do something that is not a terminal command, you will respond with 'Command not recognized'.",
    ),
    ("human", "ls"),
]
ai_msg = model.invoke(messages)
ai_msg.content
```

#### Common Model Switching with `init_chat_model`

```python
from langchain.chat_models import init_chat_model
```

```python
#with and without provider
#model = "openrouter:nvidia/nemotron-3-ultra-550b-a55b:free"
model= "nvidia/nemotron-3-ultra-550b-a55b:free"
provider = "openrouter"
#llm =  init_chat_model(model=model, temperature=0, max_tokens=1024, max_retries=2)
llm =  init_chat_model(model=model, model_provider=provider, temperature=0, max_tokens=1024, max_retries=2)
```

```python
res=llm.invoke("what is hello World")
print(res.content)
```

```python
#fetching from env by making it global variable 
import os
model= os.getenv("MODEL","openai/gpt-oss-20b")
provider = os.getenv("PROVIDER","groq")
llm =  init_chat_model(model=model, model_provider=provider, temperature=0, max_tokens=1024, max_retries=2)
```
